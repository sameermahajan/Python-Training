## list

### indexing & slicing

indexing

```python
print(l[2])
l[2] = 100
print(l)
```

slicing

```python
print(l)
print(l[-2:])
l[-2:] = [10]
print(l)
```

List Methods

```python
l.append(4)
print(l)
l.insert(0,0)
print(l)
l.extend([5,6])
print(l)
l.pop()
print(l)
l.remove(3)
print(l)
l.reverse()
print(l)
l.sort()
print(l)
print(len(l))
```

### list comprehensions

```python
nums = [1, 2, 3, 4]
squares = [n*n for n in nums]
print(squares)
```

```python
nums = [1, 2, 3, 4, 5, 6]
evens = [n for n in nums if n % 2 == 0]
print(evens)
```

## tuple

```python
# X, Y coordinates
coords = (10, 20)
```

```python
from collections import namedtuple
Point = namedtuple("Point", ["x","y"])
p = Point(3, 4)  # p.x, p.y
print(p)
print(p.x, p.y)
```

## set

```python
# dedupe and membership
items = [1,2,2,3,4,4]
unique = set(items)
print(unique)

# set operations
a = {1,2,3}
b = {3,4,5}
union = a | b
inter = a & b
diff = a - b
sym_diff = a ^ b
print(union, inter, diff, sym_diff)
# membership fast
if 3 in a:
    print("3 is in a")
```

## dict

```python
person = { "first_name": "Sameer", "last_name": "Mahajan", "Age" : 54 }
print(person)
print(person["first_name"])
person["Age"] += 1
print(person)
```

methods

```python
print(person.get("first_name"))
print(person.get("address"))
print(person.items())
print(person.keys())
print(person.values())
```

dict comprehensions

```python
fruit_basket = ["apple", "banana", "apple", "cherry"]
counts = {w: fruit_basket.count(w) for w in set(fruit_basket)}
print(counts)
```

## collections module

specialized container types

defaultdict

```python
from collections import defaultdict

d = defaultdict(list)
d["a"].append(1)
d["a"].append(2)
print(d)
```

Counter

```python
from collections import Counter

c = Counter("banana")
print(c)
print(c.most_common(1))
```

deque

```python
from collections import deque

q = deque([1,2,3])
q.appendleft(0)
q.append(4)
print(q)
```
