## 2️⃣ Nested Loops
A loop inside another loop is called a **nested loop**. It’s often used for creating patterns or working with grids.


```python
# Example: Nested loops demonstration
for i in range(3):
    for j in range(4):
        print('*', end='')
    print()
```

## 3️⃣ Loop Control Statements
- `break` → stop the loop early
- `continue` → skip the current iteration
- `pass` → do nothing (used as a placeholder)


```python
for i in range(5):
  if i == 3:
    break
  print(i)
```

```python
for i in range(5):
  if i == 3:
    continue
  print(i)
```

else statement

```python
for i in range(5):
  if i == 3:
    break
  print(i)
else:
  print('Loop finished')
```

```python
for i in range(5):
  if i == 3:
    continue
  print(i)
else:
  print('Loop finished')
```

## 6️⃣ Mini Project – Python Art Studio 🎨
Create a menu-driven program that lets users choose which pattern to print.
Use `if` statements and loops together.


```python
print("Welcome to Python Art Studio")
print("1. Right Angled Triangle")
print("2. Inverted Triangle")
print("3. Number Pyramid")
print("4. Quit")
choice = int(input("Enter your choice: "))
while choice != 4:
  if choice == 1:
    for i in range(1,20):
      print('*' * i)
  elif choice == 2:
    for i in range(20):
      print('*' * (20 - i))
  choice = int(input("Enter your choice: "))
```
