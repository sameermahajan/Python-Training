# Iterable

an object that can return an iterator

```python
from collections.abc import Iterable

print(isinstance([1,2,3], Iterable))
print(isinstance(1, Iterable))
```

# Iterator

```python
from collections.abc import Iterator

print(isinstance(iter([1,2,3]), Iterator))
print(isinstance([1,2,3], Iterator))
```

## Manual Iteration

```python
nums = [10, 20, 30]

it = iter(nums)

print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

for loop does this implicitly

```python
it = iter([10,20,30])
while True:
    try:
        item = next(it)
        print (item)
    except StopIteration:
        break
```

## Custom Iterator

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val
```

```python
for x in Countdown(5):
    print(x)
```

## Generator

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
```

```python
for x in countdown(5):
    print(x)
```

Expressions

```python
gen = (x*x for x in range(5))
```

```python
print(next(gen))    # 0
print(next(gen))    # 1
```
