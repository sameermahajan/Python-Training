## Multiple Use

Call the same function with different values for the parameters

```python
import math

def area(shape, a, b=0):
    """Calculate area of different shapes."""
    if shape == "square":
        return a * a
    elif shape == "rectangle":
        return a * b
    elif shape == "circle":
        return math.pi * (a ** 2)
    else:
        return "Shape not supported"

# 🔁 Multiple uses of the same function
print("Square area:", area("square", 5))
print("Rectangle area:", area("rectangle", 4, 6))
print("Circle area:", round(area("circle", 3), 2))
```

# Nested Functions

get_message() is defined inside greet_user().

It can access variables (name) from the outer function.

It’s not accessible outside — you can’t call get_message() directly.

```python
def greet_user(name):
    def get_message():
        return "Hello, " + name + "!"

    message = get_message()   # calling the inner function
    print(message)

# Calling the outer function
greet_user("Sameer")
```

## Type safe functions

```python
def add_numbers(a: int, b: int) -> int:
    """Add two integers and return their sum."""
    return a + b

# ✅ Correct use
print(add_numbers(5, 10))   # 15
```

```python
# ⚠️ Wrong use (will not crash, but IDEs and linters will warn)
print(add_numbers("5", 10))  # Type checker would flag this
```

## Returning Multiple Values

```python
def get_student_scores():
    """Return marks in three subjects."""
    math = 85
    science = 90
    english = 88
    return math, science, english  # returning multiple values

# Unpack the values
m, s, e = get_student_scores()

print("Math:", m)
print("Science:", s)
print("English:", e)
```

## Default values of parameters

```python
def compund_interest(principal, time, rate = 0.1):
  amount = principal * (1 + rate / (100)) ** (time)
  compound_interest = amount - principal
  return compound_interest
```

```python
compund_interest(1000,2)
```

## Named parameters

```python
compund_interest(time = 2, principal = 1000, rate = 0.2)
```
