# File Handling

open

```python
f = open("file.txt", "r")
content = f.read()
print(content)
f.close()
```

```python
with open("file.txt", "r") as f:
    text = f.read()
    print(text)
```

read line by line

```python
with open("multiline_file.txt") as f:
    for line in f:
        print(line.strip())
```

read as a list of lines

```python
with open("multiline_file.txt") as f:
    lines = f.readlines()
print (lines)
```

overwrite

```python
with open("output.txt", "w") as f:
    f.write("Hello world\n")
    f.write("Another line\n")
```

append

```python
with open("output.txt", "a") as f:
    f.write("append line\n")
```

Exception Handling

```python
try:
    with open("config.txt") as f:
      print(f.read())
except FileNotFoundError:
    print("File not found!")
```

## Binary Files

```python
import pickle
l = [1, 2]
b = open("binary.bin", "wb")
pickle.dump(l, b)
b.close()

r = open("binary.bin", "rb")
data = pickle.load(r)
r.close()
print(data)
```
