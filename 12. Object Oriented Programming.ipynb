# Session 12 - Object Oriented Programming

## Classes & Objects

```python
class Person:
    def __init__(self, name, age):
        self.name = name        # instance variable
        self.age = age

    def greet(self):
        return f"Hello, I'm {self.name}."

p1 = Person("Alice", 30)
p2 = Person("Bob", 25)

print(p1.greet())
print(p2.age)
```

## Class Variables

shared across instances

```python
class Dog:
    species = "Canis familiaris"   # class variable, shared across instances

    def __init__(self, name):
        self.name = name          # instance variable

d1 = Dog("Bruno")
d2 = Dog("Max")

print(d1.species, d2.species)
```

## Encapsulation

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner            # public
        self._balance = balance       # protected convention
        self.__secret = 12345         # private (name mangling)

    def deposit(self, amount):
        self._balance += amount

    @property
    def secret(self):
        return self.__secret

    @secret.setter
    def secret(self, secret):
        if secret < 0:
            raise ValueError("Secret cannot be negative.")
        self.__secret = secret

acc = Account("Sameer", 100)
print(acc._balance)       # allowed but discouraged
```

```python
print(acc.__secret)     # AttributeError
```

```python
print(acc._Account__secret)  # name mangling: deliberate access works
```

getters & setters

```python
print (acc.secret)
acc.secret = 54321
print (acc.secret)
```

## Inheritance & Method Override

```python
class Animal:
    def speak(self):
        return "Some sound"

class Cat(Animal):
    def speak(self):               # overriding
        return "Meow"

class Dog(Animal):
    def speak(self):
        return "Woof"
```

```python
a = Animal()
c = Cat()
d = Dog()
print(a.speak())
print(c.speak())
print(d.speak())
```

accessing inherited base class: super()

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)     # call parent constructor
        self.model = model

c = Car("Toyota", "Corolla")
print(c.brand, c.model)
```

## Special Methods

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):      # user-friendly
        return f"({self.x}, {self.y})"

    def __repr__(self):     # developer-friendly
        return f"Point(x={self.x}, y={self.y})"

p = Point(3, 4)
print(str(p))
print(repr(p))
```

addition

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = v1 + v2
print(v3.x, v3.y)
```
