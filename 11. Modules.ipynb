Using a package

```python
from mypackage.math_ops import add
from mypackage import string_ops

print(add(2, 3))
print(string_ops.shout("hello"))
```

## Import System

search path

```python
import sys
for p in sys.path:
    print(p)
```

Module Caching

```python
import utils
import utils   # second import does NOT reload module

import sys
print("cached modules:", "utils" in sys.modules)
```

reloading

```python
from importlib import reload
reload(utils)
```
