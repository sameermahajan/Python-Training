Positional & Keyword arguments

```python
def calculator(a, b, /, operator, *, round_result=False):
    if operator == "+":
        result = a + b
    else:
        result = a - b
    return round(result) if round_result else result
```

```python
calculator(10, 20, operator="+", round_result=True)
```

```python
calculator(a=10, b=20, operator="+")  # ❌ 'a' and 'b' must be positional
```

```python
calculator(10, 20, "+", True)         # ❌ round_result must be keyword
```

Positional, Keyword and Variable arguments

```python
def demo(a, b, /, c, d=10, *args, e, f=20, **kwargs):
    print("positional-only:", a, b)
    print("positional-or-keyword:", c, d)
    print("var-positional (*args):", args)
    print("keyword-only:", e, f)
    print("var-keyword (**kwargs):", kwargs)
```

```python
demo(
    1, 2,         # positional-only
    3,            # c
    4,            # d
    5, 6, 7,      # *args
    e=100, f=200, # keyword-only
    name="sameer", city="pune"   # **kwargs
)
```

## lambda

Anonymous, single-expression function

```python
# sort tuples by first element

pairs = [(1, 'a'), (3, 'c'), (2, 'b')]
sorted(pairs, key=lambda x: x[1])
```

*args

```python
# add numbers

def add(*nums):
    return sum(nums)

print(add(1,2,3,4))   # 10
print(add(1,2,3,4,5)) # 15
```

**kwargs

```python
def show_info(**data):
    for key, value in data.items():
        print(key, "->", value)

show_info(name="Sameer", age=20)
print("=" * 20)
show_info(name="Sameer", age=20, city="Pune")
```

## Higher order Functions

function argument

```python
def apply(func, x):
    return func(x)

print(apply(lambda n: n*n, 6))
```

Function returning function

```python
def multiplier(factor):
    def inner(x):
        return x * factor
    return inner

double = multiplier(2)
print(double(10))  # 20
```

Closures

inner function uses value from outer function

value is remembered even after outer function completes

```python
def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc

c = make_counter()
print(c())  # 1
print(c())  # 2
```

## Decorators

Add extra behavior to functions

Without modifying the original code

Logging, timing, authentication, caching

```python
def greet():
    print("Hello!")

def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

greet = decorator(greet)
greet()
```

@ syntax

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with {args} and {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@logger
def add(a, b):
    return a + b

print(add(3, 4))
```

with arguments

```python
def repeat(times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(times):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    print(f"Hello, {name}!")

greet("Sameer")
```

## Function Annotations

```python
def add(a: int, b: int) -> int:
    return a + b

print(add.__annotations__)
```
