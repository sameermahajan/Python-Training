## Conditions & Decisions

```python
age = int(input("Enter your age: "))
if age < 13:
    print("You are a child.")
elif age < 18:
    print("You are a teenager.")
else:
    print("You are an adult.")
```

## Loops

```python
# For loop
for i in range(5):
    print("Hello", i)

# While loop
count = 1
while count <= 5:
    print("Count:", count)
    count += 1
```
